## Audit Summary

### Dataset
- 1,470 employee records
- 35 columns
- No missing values
- No duplicate rows

### Target
- `Attrition` is the target variable
- 83.88% No
- 16.12% Yes
- The target is imbalanced

### Columns removed
- `EmployeeNumber` — identifier
- `EmployeeCount` — constant
- `Over18` — constant
- `StandardHours` — constant

### Feature groups
- Numerical features
- Ordinal features
- Nominal categorical features

### Initial observations
- Employees who left are younger on average.
- Employees who left have lower average and median monthly income.
- Employees working overtime have substantially higher observed attrition.
- These are associations, not causal conclusions.
